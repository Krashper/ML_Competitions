```python
result_data = pd.read_csv("result_data.csv")

result_data
```

```python
result_data["DateTime"] = pd.to_datetime(result_data["DateTime"])
```

```python
def get_season(row_date):
    month = row_date.month
    if month in [12, 1, 2]:
        season = "Зима"
    elif month in [3, 4, 5]:
        season = "Весна"
    elif month in [6, 7, 8]:
        season = "Лето"
    else:
        season = "Осень"
    return season
```

```python
result_data["season"] = result_data["DateTime"].apply(get_season)

result_data
```

```python
result_data["hour"] = result_data["DateTime"].apply(lambda x: x.hour)
```
