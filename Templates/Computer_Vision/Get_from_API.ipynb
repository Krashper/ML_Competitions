# Загрузка библиотек

```python
import pandas as pd
import numpy as np
import os
import requests
```

# Пример загрузки файла из API

```python
base_dir = "images" # Базовая директория

url = "https://i.pinimg.com/originals/55/9a/a0/559aa09d08a4e46e86bef50105eccf75.jpg"
response = requests.get(url)

if response.status_code == 200:
    filename = url.split("/")[-1]
    with open(os.path.join(base_dir, filename), "wb") as f:
        f.write(response.content)
    print("Файл успешно загружен")
else:
    print("Не удаётся подключиться к API:", response.status_code)
```
